```python
import pandas as pd 
messages = ( 
    pd.read_csv('archive/spam.csv', encoding='latin-1')
    .rename(columns={ 'v1': 'lable' , 'v2': 'message'})
    [['lable', 'message']]
)

messages.head()
```

```python
import nltk 
import re 
from nltk.corpus import stopwords
from nltk.stem import PorterStemmer
ps = PorterStemmer()
nltk.download('stopwords')
```

```python
corpus=[]
for i in range(0,len(messages)):
    review=re.sub('[^a-zA-z]',' ',messages['message'][i])
    review=review.lower()
    review=review.split()
    review=[ps.stem(word) for word in review if not word in stopwords.words('english')]
    review=' '.join(review)
    corpus.append(review)
corpus
```

```python
## Create the Bag of words model with the N-gram  
from sklearn.feature_extraction.text import CountVectorizer
## -- For binnay bag of words enalble binary = True , generally it is false
cv = CountVectorizer( max_features=200, binary=True)
X = cv.fit_transform(corpus).toarray()  
X.shape
```

```python
## Create the Bag of words model with the N-gram  
from sklearn.feature_extraction.text import CountVectorizer
## -- For binnay bag of words enalble binary = True , generally it is false
cv = CountVectorizer( max_features=5000, binary=True, ngram_range=(3,3))
X = cv.fit_transform(corpus).toarray()  
cv.vocabulary_
```
